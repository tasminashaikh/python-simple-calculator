```python
# Input value
value_one = float(input("Enter the first value "))
value_two = float(input("Enter the second value "))
# Calculator operations
addition_result = value_one + value_two
subtraction_result = value_one - value_two
multiplication_result = value_one * value_two
division_result = value_one / value_two
# Print final answers
print("/n====== Final Answers ======")
print("Addition Result:",addition_result)
print("Subtraction Result:",subtraction_result)
print("Multiplication Result:",multiplication_result)
print("Division Result:",division_result)
```
